```python
# ==========================================
# 🧬 PART 1: Model Setup & LoRA Injection
# ==========================================

import torch
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM, BitsAndBytesConfig
from peft import LoraConfig, get_peft_model, TaskType, prepare_model_for_kbit_training
from google.colab import drive
import os


MODEL_CHECKPOINT = "facebook/nllb-200-distilled-600M"
DEVICE = "cuda" if torch.cuda.is_available() else "cpu"

print(f"⚙️ Setting up Model on {DEVICE}...")


tokenizer = AutoTokenizer.from_pretrained(MODEL_CHECKPOINT)


bnb_config = BitsAndBytesConfig(
    load_in_4bit=True,
    bnb_4bit_use_double_quant=True,
    bnb_4bit_quant_type="nf4",
    bnb_4bit_compute_dtype=torch.float16
)


print("📥 Loading Base Model (Frozen)...")
model = AutoModelForSeq2SeqLM.from_pretrained(
    MODEL_CHECKPOINT,
    quantization_config=bnb_config,
    device_map="auto"
)


model = prepare_model_for_kbit_training(model)


lora_config = LoraConfig(
    r=32,          
    lora_alpha=64, 
    target_modules=["q_proj", "v_proj", "k_proj", "o_proj"], 
    lora_dropout=0.05,
    bias="none",
    task_type=TaskType.SEQ_2_SEQ_LM
)


print("💉 Injecting LoRA Adapters...")
model = get_peft_model(model, lora_config)


print("\n📊 Injection Report:")
model.print_trainable_parameters()
print("\n✅ PART 1 DONE: Model is injected and ready for training data.")
```

```python
# ==========================================
# 🧬 PART 1: Aggressive Model Setup & Injection
# Settings: r=64, Alpha=128, All Linear Layers
# ==========================================

import torch
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM, BitsAndBytesConfig
from peft import LoraConfig, get_peft_model, TaskType, prepare_model_for_kbit_training
from google.colab import drive
import os

# 1. إعدادات المسارات
MODEL_CHECKPOINT = "facebook/nllb-200-distilled-600M"
DEVICE = "cuda" if torch.cuda.is_available() else "cpu"

print(f"⚙️ Setting up Aggressive Model on {DEVICE}...")


tokenizer = AutoTokenizer.from_pretrained(MODEL_CHECKPOINT)


bnb_config = BitsAndBytesConfig(
    load_in_4bit=True,
    bnb_4bit_use_double_quant=True,
    bnb_4bit_quant_type="nf4",
    bnb_4bit_compute_dtype=torch.float16
)


print("📥 Loading Base Model (Frozen)...")
model = AutoModelForSeq2SeqLM.from_pretrained(
    MODEL_CHECKPOINT,
    quantization_config=bnb_config,
    device_map="auto"
)


model = prepare_model_for_kbit_training(model)


lora_config = LoraConfig(
    r=64,           
    lora_alpha=128, 
    target_modules=["q_proj", "v_proj", "k_proj", "out_proj", "fc1", "fc2"],
    lora_dropout=0.05,
    bias="none",
    task_type=TaskType.SEQ_2_SEQ_LM
)


print("💉 Injecting Huge LoRA Adapters...")
model = get_peft_model(model, lora_config)


print("\n📊 Aggressive Injection Report:")
model.print_trainable_parameters()
print("\n✅ PART 1 DONE: Model is super-charged and ready for training.")
```

```python
# ==========================================
#  PART 2: Aggressive Training Execution
# Settings: 10 Epochs, Save Best Model
# ==========================================

from transformers import Seq2SeqTrainingArguments, Seq2SeqTrainer, DataCollatorForSeq2Seq
from datasets import load_dataset


TRAIN_DATA_PATH = ''

OUTPUT_DIR = ''

print("⚙️ Preparing Data for Aggressive Training...")


dataset = load_dataset('csv', data_files=TRAIN_DATA_PATH)

dataset = dataset['train'].train_test_split(test_size=0.1)

def preprocess_function(examples):
    inputs = [str(ex) for ex in examples["en"]]
    targets = [str(ex) for ex in examples["ar"]]
    tokenizer.src_lang = "eng_Latn"
   
    model_inputs = tokenizer(inputs, text_target=targets, max_length=128, truncation=True)
    return model_inputs

tokenized_datasets = dataset.map(preprocess_function, batched=True)


training_args = Seq2SeqTrainingArguments(
    output_dir=OUTPUT_DIR,
    per_device_train_batch_size=16,  
    gradient_accumulation_steps=4,  
    learning_rate=3e-4,             
    num_train_epochs=10,            
    logging_steps=50,               
    save_strategy="epoch",          
    eval_strategy="epoch",          
    save_total_limit=2,              
    load_best_model_at_end=True,     
    metric_for_best_model="eval_loss",
    fp16=True,                       
    report_to="none"
)


trainer = Seq2SeqTrainer(
    model=model,                 
    args=training_args,
    train_dataset=tokenized_datasets["train"],
    eval_dataset=tokenized_datasets["test"],
    data_collator=DataCollatorForSeq2Seq(tokenizer, model=model),
)


print("\n🚀 Starting AGGRESSIVE Training Loop (Targeting 10 Epochs)...")
print("☕ This might take 45-60 minutes. Grab a coffee!")
trainer.train()


print("\n💾 Saving Best Model & Tokenizer...")
model.save_pretrained(OUTPUT_DIR)
tokenizer.save_pretrained(OUTPUT_DIR)
print(f"🎉 SUCCESS! Aggressive Model saved to: {OUTPUT_DIR}")
```

```python
# ==========================================
#  PART 1: Generation & Standard Metrics (En -> Ar)
# ==========================================

import torch
from peft import PeftModel, PeftConfig
from transformers import AutoModelForSeq2SeqLM, AutoTokenizer, BitsAndBytesConfig
import pandas as pd
from tqdm import tqdm
import os
import gc
import evaluate
import matplotlib.pyplot as plt 


print("🧹 Cleaning Memory...")
torch.cuda.empty_cache()
gc.collect()


ADAPTER_PATH = ''
TEST_DATA_PATH = ''
RESULTS_DIR = ''


TEMP_OUTPUT_FILE = 'temp_generated_translations.csv'
CSV_NAME = 'results_standard_finetuned.csv'
CURVE_IMG_NAME = 'chart_curve_standard_finetuned.png' 
BAR_IMG_NAME = 'chart_bar_standard_finetuned.png'    

device = "cuda" if torch.cuda.is_available() else "cpu"
print(f"⚙️ Setting up Generation on {device}...")


print("📏 Loading BLEU, METEOR, TER...")
metric_bleu = evaluate.load("sacrebleu")
metric_meteor = evaluate.load("meteor")
metric_ter = evaluate.load("ter")


print("📥 Loading Model...")
config = PeftConfig.from_pretrained(ADAPTER_PATH)
bnb_config = BitsAndBytesConfig(
    load_in_4bit=True,
    bnb_4bit_use_double_quant=True,
    bnb_4bit_quant_type="nf4",
    bnb_4bit_compute_dtype=torch.float16
)
base_model = AutoModelForSeq2SeqLM.from_pretrained(
    config.base_model_name_or_path,
    quantization_config=bnb_config,
    device_map="auto"
)
model = PeftModel.from_pretrained(base_model, ADAPTER_PATH)
tokenizer = AutoTokenizer.from_pretrained(config.base_model_name_or_path)


def translate_batch(texts, batch_size=16):
    translations = []
    target_lang_id = tokenizer.convert_tokens_to_ids("ara_Arab")
    for i in tqdm(range(0, len(texts), batch_size), desc="Translating...", leave=False):
        batch = texts[i:i+batch_size]
        tokenizer.src_lang = "eng_Latn"
        inputs = tokenizer(batch, return_tensors="pt", padding=True, truncation=True, max_length=128).to(device)
        with torch.no_grad():
            generated_tokens = model.generate(**inputs, forced_bos_token_id=target_lang_id, max_length=128)
        decoded = tokenizer.batch_decode(generated_tokens, skip_special_tokens=True)
        translations.extend(decoded)
    return translations


if os.path.exists(TEST_DATA_PATH):
    df = pd.read_csv(TEST_DATA_PATH)
    df.columns = df.columns.str.strip().str.lower()
    if 'er' in df.columns: df.rename(columns={'er': 'ar'}, inplace=True)
    df['en'] = df['en'].astype(str); df['ar'] = df['ar'].astype(str)

    unique_folds = sorted(df['fold'].unique())
    results_list = []

    
    all_predictions_data = []

    print("\n🚀 Starting Generation Loop...")

    for fold_id in unique_folds:
        fold_data = df[df['fold'] == fold_id]
        X_test = fold_data['en'].tolist()
        y_test = fold_data['ar'].tolist()

    
        preds = translate_batch(X_test)

     
        for p, r in zip(preds, y_test):
            all_predictions_data.append({'fold': fold_id, 'prediction': p, 'reference': r})

     
        score_bleu = metric_bleu.compute(predictions=preds, references=y_test)['score']
        score_meteor = metric_meteor.compute(predictions=preds, references=y_test)['meteor'] * 100
        score_ter = metric_ter.compute(predictions=preds, references=y_test)['score']

        print(f"   🌟 Fold {fold_id}: BLEU={score_bleu:.2f} | METEOR={score_meteor:.2f}")
        results_list.append({"Fold": fold_id, "BLEU": score_bleu, "METEOR": score_meteor, "TER": score_ter})

    
    df_res = pd.DataFrame(results_list)
    df_res.to_csv(os.path.join(RESULTS_DIR, CSV_NAME), index=False)

   
    df_preds = pd.DataFrame(all_predictions_data)
    df_preds.to_csv(os.path.join(RESULTS_DIR, TEMP_OUTPUT_FILE), index=False)

    # ==========================================
    # (Visualization)
    # ==========================================
    print("\n🎨 Generating Charts...")

    
    plt.figure(figsize=(10, 6))
    plt.plot(df_res['Fold'], df_res['BLEU'], marker='o', label='BLEU', linewidth=2, color='#1f77b4')
    plt.plot(df_res['Fold'], df_res['METEOR'], marker='s', label='METEOR', linewidth=2, color='#ff7f0e')
    plt.plot(df_res['Fold'], df_res['TER'], marker='x', label='TER', linestyle='--', color='#d62728', linewidth=2)
    plt.title('Standard Metrics per Fold (Fine-Tuned Model)', fontsize=14, fontweight='bold')
    plt.xlabel('Fold Number')
    plt.ylabel('Score')
    plt.legend()
    plt.grid(True, alpha=0.5)
    plt.xticks(unique_folds) 
    plt.savefig(os.path.join(RESULTS_DIR, CURVE_IMG_NAME))
    plt.close()


    avg_scores = df_res[['BLEU', 'METEOR', 'TER']].mean()

    plt.figure(figsize=(8, 6))
    metrics = ['BLEU', 'METEOR', 'TER']
    values = [avg_scores['BLEU'], avg_scores['METEOR'], avg_scores['TER']]

    bars = plt.bar(metrics, values, color=['#4CAF50', '#2196F3', '#F44336'], edgecolor='black', width=0.6)
    plt.bar_label(bars, fmt='%.2f', fontweight='bold', fontsize=12)

    plt.title('Average Standard Scores (Fine-Tuned Model)', fontsize=14, fontweight='bold')
    plt.ylabel('Average Score')
    plt.savefig(os.path.join(RESULTS_DIR, BAR_IMG_NAME))
    plt.close()

    print(f"✅ PART 1 DONE!")
    print(f"📄 CSV Saved: {CSV_NAME}")
    print(f"📈 Curve Chart Saved: {CURVE_IMG_NAME}")
    print(f"📊 Bar Chart Saved: {BAR_IMG_NAME}")
    print(f"📂 Translation Data Saved: {TEMP_OUTPUT_FILE}")

else:
    print("❌ Test file not found!")
```

```python
# ==========================================
# PART 2: BERTScore Calculation (Standalone)
# With Visualization (Curve & Bar) 📊
# ==========================================

import pandas as pd
import numpy as np
import os
from bert_score import BERTScorer
import matplotlib.pyplot as plt


RESULTS_DIR = ''
TEMP_OUTPUT_FILE = '' 
FINAL_RESULTS_FILE = ''


CURVE_IMG_NAME = 'chart_curve_bertscore_finetuned.png'
BAR_IMG_NAME = 'chart_bar_bertscore_finetuned.png'

file_path = os.path.join(RESULTS_DIR, TEMP_OUTPUT_FILE)

if os.path.exists(file_path):
    print("📂 Loading translations from Part 1...")
    df = pd.read_csv(file_path)

    print("📏 Initializing BERTScore (this might take a minute)...")

    scorer = BERTScorer(lang="ar", rescale_with_baseline=False)

  
    if scorer._tokenizer.pad_token is None:
        scorer._tokenizer.pad_token = scorer._tokenizer.eos_token
        print("🔧 Patch applied: Fixed tokenizer padding token.")

    results_list = []
    unique_folds = sorted(df['fold'].unique())

    print("\n🚀 Calculating BERTScore...")

    for fold_id in unique_folds:
        fold_data = df[df['fold'] == fold_id]
        preds = fold_data['prediction'].astype(str).tolist()
        refs = fold_data['reference'].astype(str).tolist()

        # الحساب
        P, R, F1 = scorer.score(preds, refs)
        score_f1 = F1.mean().item() * 100

        print(f"   🧠 Fold {fold_id}: BERTScore F1 = {score_f1:.2f}%")
        results_list.append({"Fold": fold_id, "BERTScore_F1": score_f1})


    df_res = pd.DataFrame(results_list)
    df_res.to_csv(os.path.join(RESULTS_DIR, FINAL_RESULTS_FILE), index=False)

    # ==========================================
    #(Visualization)
    # ==========================================
    print("\n🎨 Generating Charts...")

  
    plt.figure(figsize=(10, 6))
    plt.plot(df_res['Fold'], df_res['BERTScore_F1'], marker='o', color='#9C27B0', linewidth=2.5, label='BERTScore F1')

  
    plt.title('BERTScore Performance per Fold (Fine-Tuned)', fontsize=14, fontweight='bold')
    plt.xlabel('Fold Number', fontsize=12)
    plt.ylabel('F1 Score (%)', fontsize=12)
    plt.grid(True, alpha=0.3, linestyle='--')
    plt.xticks(unique_folds)
    plt.legend()

    
    for i, txt in enumerate(df_res['BERTScore_F1']):
        plt.text(df_res['Fold'][i], txt + 0.3, f"{txt:.1f}%", ha='center', fontweight='bold', color='#9C27B0')

    plt.savefig(os.path.join(RESULTS_DIR, CURVE_IMG_NAME))
    plt.close()

  
    avg_score = df_res['BERTScore_F1'].mean()

    plt.figure(figsize=(6, 6))
    bars = plt.bar(['Average BERTScore'], [avg_score], color='#673AB7', edgecolor='black', width=0.5)
    plt.bar_label(bars, fmt='%.2f%%', fontweight='bold', fontsize=14, padding=5)

    plt.title('Average BERTScore (Fine-Tuned)', fontsize=14, fontweight='bold')
    plt.ylabel('Score (%)')
    plt.ylim(0, 100) 

    plt.savefig(os.path.join(RESULTS_DIR, BAR_IMG_NAME))
    plt.close()

    print(f"\n✅ PART 2 DONE!")
    print(f"📄 CSV Saved: {FINAL_RESULTS_FILE}")
    print(f"📈 Curve Chart Saved: {CURVE_IMG_NAME}")
    print(f"📊 Bar Chart Saved: {BAR_IMG_NAME}")

else:
    print(f"❌ Error: Could not find '{TEMP_OUTPUT_FILE}'. Please run Part 1 first.")
```
